```python
import numpy as np
```

```python
def sigmoid(x):
    return 1 / (1 + np.exp(-x))
```

Input layer: 2 neurons
Hidden layer: 3 neurons
Output layer: 1 neuron

```python

np.random.seed(42)


input_size = 5
hidden_size = 4
output_size = 1
```

```python
W1 = np.random.randn(input_size, hidden_size)

b1 = np.random.randn(hidden_size)

W2 = np.random.randn(hidden_size, output_size)

b2 = np.random.randn(output_size)
```

```python
def forward(inputs):

    # Hidden layer
    z1 = np.dot(inputs, W1) + b1
    a1 = sigmoid(z1)

    # Output layer
    z2 = np.dot(a1, W2) + b2
    output = sigmoid(z2)

    # Scale output to 0–10 range
    player_quality = output * 10

    return player_quality
```

```python
player = np.array([85, 90, 78, 88, 80])
```

```python
quality_score = forward(player)

print("Player Ratings:")
print("Speed:", player[0])
print("Jump:", player[1])
print("Shooting:", player[2])
print("Intelligence:", player[3])
print("Strength:", player[4])

print("\nPredicted Player Quality Score:", round(float(quality_score[0]), 1))
```

The MLP works as follows:

Inputs are multiplied by weights and added to biases.

The result passes through the sigmoid activation function.

Hidden layer outputs become inputs to the output layer.

The final sigmoid produces the network's prediction
