```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
```

```python
df = pd.read_csv("titanic (2).csv")

df.head()
```

```python
df["Age"] = df["Age"].fillna(df["Age"].median())
```

```python
df["Embarked"] = df["Embarked"].fillna(df["Embarked"].mode()[0])
```

```python
df = df.drop(columns=["Cabin", "Ticket", "Name", "PassengerId"])
```

```python
df = pd.get_dummies(df, columns=["Sex", "Embarked"], drop_first=True)
```

```python
X = df.drop("Survived", axis=1)
y = df["Survived"]
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)
```

```python
rf = RandomForestClassifier(random_state=42)

rf.fit(X_train, y_train)
```

```python
y_pred = rf.predict(X_test)

accuracy = accuracy_score(y_test, y_pred)

print("Accuracy:", accuracy)
```

```python
importances = rf.feature_importances_

feature_importance = pd.DataFrame({
    "Feature": X.columns,
    "Importance": importances
}).sort_values(by="Importance", ascending=False)

print(feature_importance)
```

```python
feature_importance.plot(
    x="Feature",
    y="Importance",
    kind="bar",
    legend=False
)

plt.title("Feature Importance in Random Forest")

plt.show()
```

```python
results = []

for n in [10, 50, 100]:
    for depth in [3, 5, 7, None]:

        model = RandomForestClassifier(
            n_estimators=n,
            max_depth=depth,
            random_state=42
        )

        model.fit(X_train, y_train)

        pred = model.predict(X_test)

        acc = accuracy_score(y_test, pred)

        results.append((n, depth, acc))

        print(f"n_estimators={n}, max_depth={depth}, accuracy={acc:.3f}")
```

```python
best_model = max(results, key=lambda x: x[2])

print("Best Model:")
print("n_estimators:", best_model[0])
print("max_depth:", best_model[1])
print("Accuracy:", best_model[2])
```

The Random Forest model was trained to predict passenger survival on the Titanic dataset. Feature importance analysis showed that gender was the most greatest predictor of survival, followed by passenger class and age.

Several models were tested using different values for the number of trees (n_estimators) and the maximum tree depth (max_depth). The best-performing model achieved the highest accuracy on the test dataset with the selected parameter during the combination.
