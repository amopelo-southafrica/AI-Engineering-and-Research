```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier

from sklearn.metrics import confusion_matrix
from sklearn.metrics import accuracy_score
from sklearn.metrics import precision_score
from sklearn.metrics import recall_score
from sklearn.metrics import f1_score
```

```python
df = pd.read_csv("optdigits.tes", header=None)

df.head()
```

```python
X = df.iloc[:, :-1]   
y = df.iloc[:, -1]    
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)
```

The test set is used to find out how well the model performs on unseen data and ensures that the model generalises well rather than memorising the training data.

```python
rf = RandomForestClassifier(
    n_estimators=100,
    random_state=42
)

rf.fit(X_train, y_train)
```

```python
y_pred = rf.predict(X_test)
```

```python
cm = confusion_matrix(y_test, y_pred)

print("Confusion Matrix:")
print(cm)
```

```python
misclassified = cm.sum(axis=1) - np.diag(cm)

worst_class = np.argmax(misclassified)

print("Class with highest misclassification:", worst_class)
print("Number of errors:", misclassified[worst_class])
```

```python
accuracy = accuracy_score(y_test, y_pred)

precision = precision_score(y_test, y_pred, average="macro")

recall = recall_score(y_test, y_pred, average="macro")

f1 = f1_score(y_test, y_pred, average="macro")

print("Accuracy:", accuracy)
print("Precision:", precision)
print("Recall:", recall)
print("F1-score:", f1)
```

The Random Forest classifier was trained to recognise handwritten digits using the optdigits dataset. The model achieved strong classification performance as shown by the accuracy, precision, recall, and F1-score metrics The confusion matrix indicates which digits were most appering misclassified. Digits with similar shapes may occasionally be confused by the model, leading to higher misclassification counts for those classes.
