```python
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import confusion_matrix
```

```python
df = pd.read_csv("Iris.csv")

df.head()
```

```python
df["Species"].value_counts()
```

```python
X = df[["SepalLengthCm", "SepalWidthCm", "PetalLengthCm", "PetalWidthCm"]]
```

```python
y = df["Species"].apply(lambda x: 0 if x == "Iris-setosa" else 1)
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42
)
```

```python
model = LogisticRegression()

model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
cm = confusion_matrix(y_test, y_pred)

print(cm)
```

Model Performance

The confusion matrix shows how many predictions were correct and incorrect.
If most Iris-setosa flowers are correctly classified and very few are misclassified, the model performs well.

Since Iris-setosa is very different from the other species in the dataset, the model typically achieves high precision and recall.

```python
TP = cm[0][0]
FN = cm[0][1]
FP = cm[1][0]
TN = cm[1][1]
```

```python
accuracy = (TP + TN) / (TP + TN + FP + FN)

print("Accuracy:", accuracy)
```

```python
precision = TP / (TP + FP)

print("Precision:", precision)
```

```python
recall = TP / (TP + FN)

print("Recall:", recall)
```

```python

print("Accuracy:", accuracy)
print("Precision:", precision)
print("Recall:", recall)
```

Conclusion

The manually calculated accuracy, precision, and recall values match the results obtained from scikit-learn’s built-in metrics. This confirms that the manual calculations were implemented correctly.

The logistic regression model performs well in distinguishing Iris-setosa from the other species due to the clear differences in petal and sepal measurements.
