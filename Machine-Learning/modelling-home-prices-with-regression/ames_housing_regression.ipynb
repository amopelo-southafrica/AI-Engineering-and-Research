# Predicting Sales Prices

 ## Dataset Attributes

 The dataset contains information about houses in Ames, Iowa. The data was collected by the Ames City Assessor’s Office describing 2930 property sales which occurred in Ames, Iowa between 2006 and 2010. The dataset, containing 81 variables, was compiled and published by De Cock in 2011.

 Some of the variables contained in the original dataset have been removed from the the dataset provided to you.
 The dataset provided to you contains the following variables:
* **Year_Built:** year that the house was originally constructed
* **Year_Remod_Add:** year that the house was last remodelled
* **Total_Bsmt_SF:** total size of basement area in square feet
* **First_Flr_SF:** size of the first floor in square feet
* **Second_Flr_SF:** size of the second floor in square feet
* **Gr_Liv_Area:** size of above grade, ground living area in square feet
* **Full_Bath:** number of full above grade bathrooms in the house
* **Half_Bath:** number of half above grade bathrooms in the house
* **Bedroom_AbvGr:** number of above grade bedrooms (does not include basement bedrooms)
* **Kitchen_AbvGr:** number of above grade kitchens
* **TotRms_AbvGrd:** total number of above grade rooms (does not include bathrooms)
* **Fireplaces:** number of fireplaces in the house
* **Garage_Area:** size of garage in square feet
* **Sale_Price:** sale price of the house in dollars


*De Cock, D. (2011). "Ames, Iowa: Alternative to the Boston Housing Data as an End of Semester
Regression Project," Journal of Statistics Education, Volume 19, Number 3.*

 ## Objective

 The goal of this task is to analyse the relationship between these variables and build a multiple linear regression model to predict the sales prices based on the 'Gr_Liv_Area' and 'Garage_Area` variables.


## Loading the Dataset

The first step is to load the `ames.csv` dataset into the Jupyter notebook using pandas.

This dataset contains information about houses in Ames, Iowa, including property features and sale prices.  
The goal of this project is to analyse the dataset and build a regression model that predicts the **SalePrice** of a house based on selected features.

After loading the dataset, the first few rows are displayed to confirm that the dataset has been loaded correctly.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
```

## Inspecting the Dataset

Before performing analysis, it is important to understand the structure of the dataset.

In this step we:
- Display column names
- Check data types
- Identify missing values

This helps ensure that the dataset is clean and ready for analysis.

```python
df = pd.read_csv("ames.csv")

df.head()
df.info()
df.describe()
```

```python
df.isnull().sum()
df = df.dropna()
print(df.columns)
```

## Exploratory Data Analysis

Exploratory Data Analysis (EDA) helps us understand the dataset by visualising important variables.

In this section we explore:

1. The distribution of the dependent variable (SalePrice)
2. The relationship between living area and sale price
3. The relationship between garage area and sale price

These visualisations help identify patterns and relationships that may influence house prices.

### Distribution of Sale Prices

This histogram shows the distribution of house sale prices in the dataset.

Understanding the distribution of the dependent variable helps determine whether the data is skewed and provides insight into the overall housing market in Ames.

```python
sns.histplot(df["Sale_Price"], kde=True)

plt.title("Distribution of Sale Prices")
plt.xlabel("Sale Price")
plt.ylabel("Frequency")

plt.show()
```

### Living Area vs Sale Price

This scatter plot explores the relationship between the size of the above-ground living area (`Gr_Liv_Area`) and the house sale price.

We expect houses with larger living areas to generally have higher sale prices.

```python
sns.scatterplot(x=df["Gr_Liv_Area"], y=df["Sale_Price"])

plt.title("Living Area vs Sale Price")

plt.show()
```

### Garage Area vs Sale Price

This scatter plot examines the relationship between garage size (`Garage_Area`) and sale price.

Larger garages may contribute positively to property value.

```python
sns.scatterplot(x=df["Garage_Area"], y=df["Sale_Price"])

plt.title("Garage Area vs Sale Price")

plt.show()
```

## Defining Variables

In regression analysis, we separate the dataset into:

- **Independent variables (X)** – features used to predict the target
- **Dependent variable (y)** – the value we want to predict

According to the task requirements, the model will use:

- Gr_Liv_Area
- Garage_Area

to predict **SalePrice**.

```python
X = df[["Gr_Liv_Area", "Garage_Area"]]
y = df["Sale_Price"]
```

## Splitting the Dataset

To evaluate the model fairly, the dataset is divided into:

- **Training set (75%)** – used to train the regression model
- **Testing set (25%)** – used to evaluate model performance

This ensures that the model can generalise to unseen data.

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42
)
```

## Building the Linear Regression Model

A multiple linear regression model is trained using the training dataset.

The model learns the relationship between the independent variables and the sale price.

```python
model = LinearRegression()

model.fit(X_train, y_train)
```

## Model Coefficients

The intercept and coefficients represent how each feature influences the predicted house price.

- **Intercept** – baseline value when all predictors are zero
- **Coefficients** – expected change in sale price for a one-unit increase in each feature

```python
print("Intercept:", model.intercept_)

print("Coefficients:")

for feature, coef in zip(X.columns, model.coef_):
    print(feature, coef)
```

## Generating Predictions

The trained model is used to predict house prices for the test dataset.

These predicted values will later be compared to the actual sale prices.

```python
y_pred = model.predict(X_test)
```

## Model Evaluation

To measure model performance, we calculate the Root Mean Squared Error (RMSE).

RMSE measures the average prediction error between the predicted sale prices and the actual sale prices.

```python
mse = mean_squared_error(y_test, y_pred)

rmse = np.sqrt(mse)

print("Mean Squared Error:", mse)
print("Root Mean Squared Error:", rmse)
```

## Error Plot: Predicted vs Actual Values

This scatter plot compares predicted house prices with actual house prices.

If the model performs well, the points should align closely along the diagonal line.

```python
plt.scatter(y_test, y_pred)

plt.xlabel("Actual Prices")
plt.ylabel("Predicted Prices")

plt.title("Predicted vs Actual Sale Prices")

plt.show()
```

```python

print("Intercept:", model.intercept_)

coefficients = pd.DataFrame({
    "Feature": ["Gr_Liv_Area", "Garage_Area"],
    "Coefficient": model.coef_
})

print(coefficients)



median_price = df["Sale_Price"].median()

print("Median Sale Price:", median_price)
```
