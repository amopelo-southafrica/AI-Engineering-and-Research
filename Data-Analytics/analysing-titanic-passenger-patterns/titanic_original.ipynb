# Exploratory Data Analysis of the Titanic Dataset

## Introduction
This notebook presents an exploratory data analysis (EDA) of the Titanic dataset. The aim is to understand the structure of the data, identify patterns, examine missing values, and explore which factors influenced passenger survival.

The analysis focuses on the following guiding questions:
1. What is the most important factor in determining whether a person was likely to survive?
2. Did passenger class affect access to lifeboats and survival?
3. Was the rule of “women and children first” reflected in the data?
4. What other interesting observation can be made from the dataset?

The notebook uses summary statistics and visualisations to support each conclusion.

## Step 1: Importing Libraries

In this step, the necessary Python libraries are imported.

- **pandas** is used for reading and analysing the dataset.
- **numpy** is used for numerical operations.
- **matplotlib** and **seaborn** are used to create visualisations.

These libraries provide the tools needed to explore the Titanic dataset both numerically and visually.

```python
# Import libraries for data analysis and visualization
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Make plots display inside notebook
%matplotlib inline
```

## Step 2: Loading the Dataset

The Titanic dataset is loaded into a pandas DataFrame. This allows the data to be stored in a tabular format so that it can be explored and analysed more easily.

The first few rows are displayed to understand the structure of the dataset and to identify the available variables, such as passenger class, sex, age, fare, and survival outcome.

```python
# Load Titanic dataset
df = pd.read_csv("Titanic.csv")

# Display first rows
df.head()
```

## Step 3: Inspecting the Dataset Structure

The purpose of this step is to understand the overall structure of the dataset.

By using functions such as `head()` and `info()`, it becomes possible to:
- see the columns in the dataset,
- determine the data types of each column,
- identify how many non-null values are present,
- get an initial sense of data quality.

This step is important because it helps identify which variables are numerical, which are categorical, and which may require cleaning before analysis.

```python
# View dataset information
df.info()
```

```python
# Summary statistics for numerical columns
df.describe()
```

```python
# Check for missing values
df.isnull().sum()
```

```python
print(df.columns)
```

## Step 4: Summary Statistics

Summary statistics are used to describe the numerical features of the dataset.

This includes values such as:
- mean,
- minimum,
- maximum,
- standard deviation,
- quartiles.

These statistics help to understand the distribution of variables such as age, fare, siblings/spouses aboard, and parents/children aboard. This provides a foundation for identifying trends, unusual values, and possible outliers.

```python
# Survival distribution
sns.countplot(x="Survived", data=df)

plt.title("Survival Distribution")
plt.xlabel("Survived (0 = No, 1 = Yes)")
plt.ylabel("Passenger Count")

plt.show()
```

## Step 5: Checking for Missing Values

Before carrying out deeper analysis, it is important to check for missing values.

Missing values can affect the reliability of the results and may influence visualisations and summary statistics. In the Titanic dataset, some columns such as age, cabin, and embarked may contain missing data.

Identifying missing values helps determine whether data cleaning or careful interpretation is needed in later steps.

```python
# Survival by passenger class
sns.countplot(x="Pclass", hue="Survived", data=df)

plt.title("Survival by Passenger Class")
plt.xlabel("Passenger Class")
plt.ylabel("Count")

plt.show()
```

Observation:
Passengers in first class had significantly higher survival rates compared to second and third class passengers.

## Step 6: Overall Survival Distribution

This visualisation shows the overall distribution of passengers who survived and those who did not survive.

The purpose of this step is to get a general understanding of the survival outcome before comparing it against other variables such as passenger class, sex, and age.

### Interpretation
From this graph, it is possible to observe whether survival was common or rare among passengers. This provides a baseline for the rest of the analysis.

## Step 8: Survival by Gender

This section examines whether gender influenced survival.

The phrase “women and children first” suggests that female passengers may have been prioritised during evacuation. Comparing survival by sex helps test whether this pattern appears in the dataset.

### Interpretation
If the graph shows that females had a much higher survival rate than males, then gender was a strong factor influencing survival.

### Answer to Guiding Question
The data shows that female passengers were far more likely to survive than male passengers. This strongly suggests that gender was one of the most important factors affecting survival.

```python
# Survival by gender
sns.countplot(x="Sex", hue="Survived", data=df)

plt.title("Survival by Gender")
plt.xlabel("Gender")
plt.ylabel("Passenger Count")

plt.show()
```

## Step 9: Age Distribution and the Role of Children

This section explores passenger age to determine whether children may also have been prioritised during evacuation.

Age is an important feature because the phrase “women and children first” implies that younger passengers may have had a greater chance of survival. The age distribution also helps to understand the demographic makeup of the passengers on board.

### Interpretation
If younger passengers show relatively stronger survival outcomes than adults, this would support the idea that children were prioritised.

### Answer to Guiding Question
The data suggests that children had a better chance of survival than many adults, although the effect is not as visually strong as the effect of gender. The strongest evidence in support of “women and children first” comes from the much higher survival rate among women, with children also showing some advantage.

```python
# Age distribution of passengers
sns.histplot(df["Age"].dropna(), kde=True)

plt.title("Age Distribution of Passengers")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```

## Step 10: Relationship Between Variables

A correlation heatmap is used to examine the relationships between the numerical variables in the dataset.

This helps identify whether variables such as class, fare, age, siblings/spouses aboard, and parents/children aboard are related to survival.

### Interpretation
The heatmap helps reveal which numerical features move most closely with survival, but correlation alone does not fully explain survival because important variables such as sex are categorical.

### Answer to Guiding Question
Based on the visual analysis, **sex appears to be the most important factor in determining survival**, with passenger class also having a strong influence. Female passengers had much higher survival rates, and first-class passengers survived more often than lower-class passengers.
12. Extra observation

```python
# Correlation heatmap
plt.figure(figsize=(10,6))

sns.heatmap(df.corr(numeric_only=True), annot=True, cmap="coolwarm")

plt.title("Feature Correlation Heatmap")

plt.show()
```

## Distribution of Ticket Fare

This visualization shows the distribution of ticket fares paid by passengers aboard the Titanic. 
The purpose of this analysis is to understand how ticket prices were spread across passengers and 
whether most passengers paid low or high fares.

### Interpretation
The distribution shows that most passengers paid relatively low ticket fares, while a smaller 
number of passengers paid very high fares. This indicates that the dataset is right-skewed, 
meaning there are a few high-value outliers.

Higher fares are often associated with first-class passengers, which may explain why wealthier 
passengers had better access to lifeboats and higher survival rates.

### Insight
This observation suggests that socio-economic status, reflected through ticket fare and passenger 
class, may have played a role in survival outcomes during the Titanic disaster.

```python
# Distribution of ticket fares
sns.histplot(df["Fare"], kde=True)

plt.title("Fare Distribution")
plt.xlabel("Fare")
plt.ylabel("Frequency")

plt.show()
```

## Conclusion

This exploratory data analysis examined the main factors that influenced survival on the Titanic.

The analysis showed the following:
- Survival was influenced strongly by **sex**, with women having a much higher survival rate than men.
- **Passenger class** also mattered, with first-class passengers surviving more often than lower-class passengers.
- The idea of **“women and children first”** is supported by the data, especially in the case of women.
- An additional observation is that **higher fares were associated with higher survival**, suggesting that wealth and class affected outcomes.

Overall, the data suggests that survival on the Titanic was shaped mainly by gender and social class rather than chance alone.
