```python
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler
countries = pd.read_csv('countries.csv')
countries = countries.dropna(subset=['EG.ELC.ACCS.ZS']).copy()
```

```python
plt.hist(countries["EG.ELC.ACCS.ZS"])
plt.title("Electricity Access (Original)")
plt.xlabel("Percentage")
plt.ylabel("Frequency")
plt.show()
```

```python
scaler = MinMaxScaler()

countries["scaled_electricity"] = scaler.fit_transform(
    countries[["EG.ELC.ACCS.ZS"]]
)
```

```python
fig, axes = plt.subplots(1,2, figsize=(10,4))

axes[0].hist(countries["EG.ELC.ACCS.ZS"])
axes[0].set_title("Original Data")

axes[1].hist(countries["scaled_electricity"])
axes[1].set_title("Scaled Data")

plt.show()
```

Normalisation makes more sense.
This is because the variables are measured on different scales (40 vs 10), and normalisation scales them between 0 and 1 so they can be compared.

Standardisation makes more sense.
This is because the study time values are not normally distributed, and standardisation helps handle variables that have different distributions
