```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
cars = pd.read_csv("Cars93.csv")
cars.head()
```

```python
cars.info()
cars.describe()
```

```python
manufacturers = cars[cars["Manufacturer"].isin(["Audi","Hyundai","Suzuki","Toyota"])]

sns.boxplot(x="Manufacturer", y="Rev.per.mile", data=manufacturers)

plt.title("Revs per Mile by Manufacturer")
plt.show()
```

```python
plt.plot(cars["Wheelbase"], cars["Turn.circle"], "o")

plt.xlabel("Wheelbase")
plt.ylabel("Turning Circle")
plt.title("Wheelbase vs Turning Circle")

plt.show()
```

```python
mean_hp = cars.groupby("Type")["Horsepower"].mean()

mean_hp.plot(kind="bar")

plt.title("Average Horsepower by Car Type")
plt.xlabel("Car Type")
plt.ylabel("Mean Horsepower")

plt.show()
```

```python
plt.hist(cars["MPG.city"], alpha=0.5, label="City MPG")
plt.hist(cars["MPG.highway"], alpha=0.5, label="Highway MPG")

plt.legend()
plt.title("City MPG vs Highway MPG")
plt.xlabel("Miles Per Gallon")
plt.ylabel("Frequency")

plt.show()
```
